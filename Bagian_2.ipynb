# **DEKOMPOSISI MATRIKS**

## Melakukan Dekomposisi LU

```python
import numpy as np

A = np.array([
    [2, 3, 1, 5],
    [4, 7, 7, 17],
    [-2, 4, 5, 10],
    [6, 18, 22, 61]
], dtype=float)

def lu_doolittle(A):
    n = A.shape[0]
    L = np.zeros((n, n))
    U = np.zeros((n, n))

    for i in range(n):
        # Diagonal L = 1
        L[i, i] = 1

        # Menghitung elemen U
        for j in range(i, n):
            U[i, j] = A[i, j] - sum(L[i, k] * U[k, j] for k in range(i))

        # Menghitung elemen L
        for j in range(i + 1, n):
            L[j, i] = (A[j, i] - sum(L[j, k] * U[k, i] for k in range(i))) / U[i, i]

    return L, U

L, U = lu_doolittle(A)

print("Matriks A:")
print(A)

print("\nMatriks L:")
print(L)

print("\nMatriks U:")
print(U)

A_rekonstruksi = L @ U

print("\nHasil perkalian L @ U:")
print(A_rekonstruksi)

print("\nApakah A = L @ U?")
print(np.allclose(A, A_rekonstruksi))
```

## Melakukan Dekomposisi QR

```python
import numpy as np

# ======================================================
# 1. Mendefinisikan matriks A (4x4)
# ======================================================
# Matriks ini akan didekomposisi menjadi Q dan R
A = np.array([
    [2, 3, 1, 5],
    [4, 7, 7, 17],
    [-2, 4, 5, 10],
    [6, 18, 22, 61]
], dtype=float)

print("Matriks A:")
print(A)

# ======================================================
# 2. Melakukan dekomposisi QR
# ======================================================
# Fungsi np.linalg.qr() mengembalikan:
# Q : matriks ortogonal (Q^T Q = I)
# R : matriks segitiga atas
Q, R = np.linalg.qr(A)

print("\nMatriks Q (ortogonal):")
print(Q)

print("\nMatriks R (segitiga atas):")
print(R)

# ======================================================
# 3. Verifikasi bahwa A ≈ Q @ R
# ======================================================
# Mengalikan Q dan R untuk merekonstruksi matriks A
A_rekonstruksi = Q @ R

print("\nHasil perkalian Q @ R:")
print(A_rekonstruksi)

# Mengecek apakah A dan Q@R hampir sama
# Digunakan 'allclose' karena perhitungan numerik
print("\nApakah A ≈ Q @ R?")
print(np.allclose(A, A_rekonstruksi))

# ======================================================
# 4. Verifikasi tambahan: Ortogonalitas matriks Q
# ======================================================
# Jika Q benar-benar ortogonal, maka Q^T @ Q = I
QTQ = Q.T @ Q

print("\nHasil Q^T @ Q (harus mendekati matriks identitas):")
print(QTQ)
```

## Menghitung Eigenvalue dan Eigenvector

```python
import numpy as np

# ======================================================
# 1. Mendefinisikan matriks A (4x4)
# ======================================================
A = np.array([
    [2, 3, 1, 5],
    [4, 7, 7, 17],
    [-2, 4, 5, 10],
    [6, 18, 22, 61]
], dtype=float)

print("Matriks A:")
print(A)

# ======================================================
# 2. Menghitung eigenvalue dan eigenvector
# ======================================================
# np.linalg.eig() adalah fungsi bawaan NumPy
# eigvals  : vektor eigenvalue
# eigvecs  : matriks eigenvector (setiap kolom = 1 eigenvector)
eigvals, eigvecs = np.linalg.eig(A)

print("\nEigenvalue:")
print(eigvals)

print("\nEigenvector (setiap kolom adalah eigenvector):")
print(eigvecs)

# ======================================================
# 3. Verifikasi Av = λv untuk setiap eigenvalue
# ======================================================
for i in range(len(eigvals)):
    left = A @ eigvecs[:, i]              # A v
    right = eigvals[i] * eigvecs[:, i]    # λ v

    print(f"\nVerifikasi eigen ke-{i+1}:")
    print("A v =", left)
    print("λ v =", right)
    print("Apakah Av ≈ λv ?", np.allclose(left, right))
```

## Menuliskan  Interpretasi 

```python
import numpy as np

# Matriks A
A = np.array([
    [2, 3, 1, 5],
    [4, 7, 7, 17],
    [-2, 4, 5, 10],
    [6, 18, 22, 61]
], dtype=float)

# Menghitung eigenvalue dan eigenvector
eigvals, eigvecs = np.linalg.eig(A)

print("Eigenvalue:")
print(eigvals)

print("\nEigenvector (setiap kolom adalah eigenvector):")
print(eigvecs)

# Memilih eigenvalue dan eigenvector pertama
lambda_1 = eigvals[0]
v_1 = eigvecs[:, 0]

print("\nEigenvalue terpilih (λ):")
print(lambda_1)

print("\nEigenvector terpilih (v):")
print(v_1)

# Verifikasi Av = λv
Av = A @ v_1
lambda_v = lambda_1 * v_1

print("\nA v:")
print(Av)

print("\nλ v:")
print(lambda_v)

print("\nApakah Av ≈ λv?")
print(np.allclose(Av, lambda_v))
```
