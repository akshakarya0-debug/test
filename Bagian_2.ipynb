# **DEKOMPOSISI MATRIKS**

## Melakukan Dekomposisi LU

```python
import numpy as np

A = np.array([
    [2, 3, 1, 5],
    [4, 7, 7, 17],
    [-2, 4, 5, 10],
    [6, 18, 22, 61]
], dtype=float)

def lu_doolittle(A):
    n = A.shape[0]
    L = np.zeros((n, n))
    U = np.zeros((n, n))

    for i in range(n):
        # Diagonal L = 1
        L[i, i] = 1

        # Menghitung elemen U
        for j in range(i, n):
            U[i, j] = A[i, j] - sum(L[i, k] * U[k, j] for k in range(i))

        # Menghitung elemen L
        for j in range(i + 1, n):
            L[j, i] = (A[j, i] - sum(L[j, k] * U[k, i] for k in range(i))) / U[i, i]

    return L, U

L, U = lu_doolittle(A)

print("Matriks A:")
print(A)

print("\nMatriks L:")
print(L)

print("\nMatriks U:")
print(U)

A_rekonstruksi = L @ U

print("\nHasil perkalian L @ U:")
print(A_rekonstruksi)

print("\nApakah A = L @ U?")
print(np.allclose(A, A_rekonstruksi))
```

## Melakukan Dekomposisi QR

```python
import numpy as np

# ======================================================
# 1. Mendefinisikan matriks A (4x4)
# ======================================================
# Matriks ini akan didekomposisi menjadi Q dan R
A = np.array([
    [2, 3, 1, 5],
    [4, 7, 7, 17],
    [-2, 4, 5, 10],
    [6, 18, 22, 61]
], dtype=float)

print("Matriks A:")
print(A)

# ======================================================
# 2. Melakukan dekomposisi QR
# ======================================================
# Fungsi np.linalg.qr() mengembalikan:
# Q : matriks ortogonal (Q^T Q = I)
# R : matriks segitiga atas
Q, R = np.linalg.qr(A)

print("\nMatriks Q (ortogonal):")
print(Q)

print("\nMatriks R (segitiga atas):")
print(R)

# ======================================================
# 3. Verifikasi bahwa A ≈ Q @ R
# ======================================================
# Mengalikan Q dan R untuk merekonstruksi matriks A
A_rekonstruksi = Q @ R

print("\nHasil perkalian Q @ R:")
print(A_rekonstruksi)

# Mengecek apakah A dan Q@R hampir sama
# Digunakan 'allclose' karena perhitungan numerik
print("\nApakah A ≈ Q @ R?")
print(np.allclose(A, A_rekonstruksi))

# ======================================================
# 4. Verifikasi tambahan: Ortogonalitas matriks Q
# ======================================================
# Jika Q benar-benar ortogonal, maka Q^T @ Q = I
QTQ = Q.T @ Q

print("\nHasil Q^T @ Q (harus mendekati matriks identitas):")
print(QTQ)
```
