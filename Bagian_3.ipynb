```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Salary_dataset.csv')
df.head()
```

```python
x = df["YearsExperience"].values
y = df["Salary"].values

X = np.column_stack((np.ones_like(x), x))

print("X =\n", X)
print("y =\n", y)
```

```python
# Menghitung parameter regresi dengan matriks (Normal Equation)
XtX = X.T @ X
XtX_inv = np.linalg.inv(XtX)
Xty = X.T @ y

theta = XtX_inv @ Xty
print("Parameter regresi (theta) =", theta)

```

```python
# Visualisasi hasil regresi
import matplotlib.pyplot as plt

plt.scatter(x, y, color='blue', label='Data Asli')
plt.plot(x, X @ theta, color='red', label='Regresi Linier')
plt.xlabel('YearsExperience')
plt.ylabel('Salary')
plt.title('Regresi Linier: Salary vs YearsExperience')
plt.legend()
plt.show()
```
